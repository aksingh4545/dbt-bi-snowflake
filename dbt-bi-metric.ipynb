### Create user and assign the role to them like i have done as TRANSFORM , that all user password ,login name , role and warehouse required in configuration with dbt local.

```
%%sql -r dataframe_9
-- Step 4: Create the `dbt` user and assign to the transform role
CREATE USER IF NOT EXISTS dbt
  PASSWORD='dbtPassword123'
  LOGIN_NAME='dbt'
  MUST_CHANGE_PASSWORD=FALSE
  DEFAULT_WAREHOUSE='COMPUTE_WH'
  DEFAULT_ROLE=TRANSFORM
  DEFAULT_NAMESPACE='dbt_bi_metrics_platform_db.RAW'
  COMMENT='DBT user used for data transformation';
ALTER USER dbt SET TYPE = LEGACY_SERVICE;
GRANT ROLE TRANSFORM TO USER dbt;
```

### Assigning permission to Role

```
%%sql -r dataframe_4
-- Step 6: Grant permissions to the `transform` role
GRANT ALL ON WAREHOUSE COMPUTE_WH TO ROLE TRANSFORM;
GRANT ALL ON DATABASE dbt_bi_metrics_platform_db TO ROLE TRANSFORM;
GRANT ALL ON ALL SCHEMAS IN DATABASE dbt_bi_metrics_platform_db TO ROLE TRANSFORM;
GRANT ALL ON FUTURE SCHEMAS IN DATABASE dbt_bi_metrics_platform_db TO ROLE TRANSFORM;
GRANT ALL ON ALL TABLES IN SCHEMA dbt_bi_metrics_platform_db.RAW TO ROLE TRANSFORM;
GRANT ALL ON FUTURE TABLES IN SCHEMA dbt_bi_metrics_platform_db.RAW TO ROLE TRANSFORM;
```
